```python
import pandas as pd
import numpy as np

data = pd.read_csv('online_retail_II.csv')
```

```python
data.dropna(subset=['Quantity','Price'],inplace=True)
data['TotalPrice'] = data['Quantity']*data['Price']
```

```python
X = data[['Quantity','Price']].values
y = data['TotalPrice'].values

X = (X - np.mean(X, axis=0))/np.std(X,axis=0)

X=np.hstack((np.ones((X.shape[0],1)),X))
```

```python
weights = np.zeros(X.shape[1])
learning_rate = 0.01
iterations = 1000
n = len(y)
```

```python
for i in range(iterations):
    predictions = np.dot(X, weights)
    errors = predictions - y
    gradient = (1/n) * np.dot(X.T, errors)
    weights -= learning_rate * gradient
    
    mse = (1/n) * np.sum(errors**2)
    if i%100 == 0:
        print(f"Iteration {i}, MSE: {mse:.2f}")
```

```python
print(f"Final Weights: {weights}")

predictions = np.dot(X, weights)
mse = (1/n) * np.sum((predictions - y)**2)
print(f"Mean Squared Error: {mse:.2f}")

```
